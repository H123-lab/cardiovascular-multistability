# Interactive Exploration of Cardiovascular Multistability
This notebook provides an interactive walkthrough of the model execution, tracking data variables from initialization states down to stable limit-cycle orbits.

```python
import numpy as np
import matplotlib.pyplot as plt
from src.model.parameters import get_baseline_parameters
from src.model.cardiovascular_model import run_rk4_integration
from src.utils import calculate_cycle_metrics
print("Data Science Stack Initialized Successfully.")
```

## 1. Execute Baseline Run ($R_{sys} = 1.0$)

```python
p = get_baseline_parameters()
history, time_array = run_rk4_integration(np.array([120.0, 95.0, 12.0]), Rsys=1.0, p=p)
metrics = calculate_cycle_metrics(history, time_array, p)
print(f"Limit Cycle Validated: Baseline MAP = {metrics['MAP']:.4f} mmHg")
```

## 2. Interactive Phase-Space Plot Visualization

```python
plt.figure(figsize=(6, 4.5))
plt.plot(history[0, -800:], history[1, -800:], 'g-', label="Steady PV Orbit")
plt.title("Left Ventricular Pressure-Volume Attractor Target Loop")
plt.xlabel("Volume (mL)")
plt.ylabel("Pressure (mmHg)")
plt.grid(True)
plt.legend()
plt.show()
```
